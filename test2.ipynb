```python

#a = np.array([1, 5, 10, 3, 4, 25, 30])

#a = np.array([11, 15, 20, 21, 35, 40, 45])
a = np.array([2, 4, 6, 8])


```

```python
def homework(a):
    # Find all elements in the array that are divisible by 5
    result1 = np.extract(a % 5 == 0, a)
    # Find all elements in result1 that are odd
    my_result = np.extract(result1 % 2 == 1, result1)
    return my_result
```

```python
# Run this cell after completing homework().
your_answer = homework(a)
print("Your answer:", your_answer)
```
