```python
# common
import numpy as np
import pandas as pd
from sklearn import linear_model

# Libraries for retrieving data from the web and handling zip files
import requests, zipfile
from io import StringIO
import io
```

```python
# Specify the url with data
url = 'https://github.com/Hernan4444/MyAnimeList-Database/archive/refs/heads/master.zip'

# Acquire data from the url
r = requests.get(url, stream=True)

# read and extract the zipfile
z = zipfile.ZipFile(io.BytesIO(r.content))
z.extractall()

# Load and clean the data
anime_data = pd.read_csv('MyAnimeList-Database-master/data/anime.csv')
anime_data_extracted = anime_data[anime_data['Score'] != 'Unknown'].copy()
anime_data_extracted['Score'] = pd.to_numeric(anime_data_extracted['Score'])
```

```python
def homework(anime_data_extracted, X_column, Y_column):
    # 1. Extract the explanatory variable (X). 
    # We use double brackets [[ ]] because scikit-learn requires X to be 2-Dimensional (a DataFrame)
    X = anime_data_extracted[[X_column]]
    
    # 2. Extract the target variable (Y).
    # We use single brackets [ ] because Y can just be 1-Dimensional (a Pandas Series)
    Y = anime_data_extracted[Y_column]
    
    # 3. Create the linear regression model object
    model = linear_model.LinearRegression()
    
    # 4. Train (fit) the model using the X and Y data
    model.fit(X, Y)
    
    # 5. Calculate the coefficient of determination (R-squared). 
    # The .score() method automatically calculates and returns this as a float.
    result = model.score(X, Y)
    
    # 6. Return the calculated float
    return result
```

```python
# Run this cell after completing homework().
your_answer = homework(
    anime_data_extracted,
    X_column="Members",
    Y_column="Completed",
)
print("Your answer:", your_answer)
```

```python
case_2_data = pd.DataFrame({
    "X": [1, 2, 3, 4],
    "Y": [2, 4, 6, 8],
})
case_2_answer = homework(case_2_data, X_column="X", Y_column="Y")
print("Your answer:", case_2_answer)
```

```python
case_3_data = pd.DataFrame({
    "X": [1, 2, 3, 4],
    "Y": [1, 2, 1, 2],
})
case_3_answer = homework(case_3_data, X_column="X", Y_column="Y")
print("Your answer:", case_3_answer)
```
